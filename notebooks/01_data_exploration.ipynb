# DriverGuard AI - Notebook 01: Dataset Exploration

This notebook performs exploratory data analysis (EDA) on the Eye State Dataset (e.g., MRL Eye Dataset).
It validates dataset structure, inspects class balance, analyzes image resolutions, and visualizes sample eye patches.

```python
import os
import glob
from pathlib import Path
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("../data/raw")
print(f"Checking dataset path: {DATA_DIR.resolve()}")
```

### 1. Dataset Directory and Class Distribution

```python
classes = ["Closed", "Open"]
counts = {}

for cls in classes:
    pattern = str(DATA_DIR / cls / "*.*")
    images = glob.glob(pattern)
    counts[cls] = len(images)
    print(f"Class '{cls}': {len(images)} images found.")

df_counts = pd.DataFrame(list(counts.items()), columns=["Class", "Count"])
plt.figure(figsize=(6, 4))
plt.bar(df_counts["Class"], df_counts["Count"], color=["#EF4444", "#10B981"])
plt.title("Class Distribution in Eye Dataset")
plt.ylabel("Number of Images")
plt.show()
```

### 2. Sample Image Inspection and CLAHE Preprocessing

```python
from src.vision.preprocessing import EyePreprocessor

preprocessor = EyePreprocessor(target_size=(64, 64), apply_clahe=True)

# Create figure to compare raw vs preprocessed samples
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
fig.suptitle("Sample Eye Patches: Raw vs. CLAHE Preprocessed (64x64)", fontsize=14)

for i, cls in enumerate(classes):
    pattern = str(DATA_DIR / cls / "*.*")
    files = glob.glob(pattern)[:4]
    for j, f in enumerate(files):
        img = cv2.imread(f)
        if img is not None:
            processed = preprocessor.preprocess_crop(img)
            axes[i, j].imshow(processed)
            axes[i, j].set_title(f"{cls} Eye #{j+1}")
            axes[i, j].axis("off")

plt.tight_layout()
plt.show()
```
