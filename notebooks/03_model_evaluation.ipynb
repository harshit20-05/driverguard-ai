# DriverGuard AI - Notebook 03: Model Evaluation & Latency Profiling

This notebook computes rigorous evaluation metrics on the held-out test set,
generates confusion matrices, plots ROC curves, and profiles real-time inference latency.

```python
import os
import time
import json
import numpy as np
import tensorflow as tf
from src.models.evaluation import compute_metrics, plot_and_save_confusion_matrix, plot_and_save_training_curves

MODEL_PATH = "../models/eye_classifier.keras"
print(f"Evaluating Model: {MODEL_PATH}")
```

### 1. Load Model and Held-Out Test Set

```python
if os.path.exists(MODEL_PATH):
    model = tf.keras.models.load_model(MODEL_PATH)
    print("Loaded model successfully.")
else:
    print("Model file not found. Ensure training has completed first.")
```

### 2. Inference Latency Benchmark

```python
if 'model' in locals():
    dummy_input = np.random.rand(2, 64, 64, 3).astype(np.float32)
    # Warm up
    for _ in range(10):
        _ = model(dummy_input, training=False)

    latencies = []
    for _ in range(100):
        t0 = time.perf_counter()
        _ = model(dummy_input, training=False)
        latencies.append((time.perf_counter() - t0) * 1000.0)

    print(f"Average Dual-Eye Inference Latency: {np.mean(latencies):.2f} ms")
    print(f"P95 Latency: {np.percentile(latencies, 95):.2f} ms")
```
